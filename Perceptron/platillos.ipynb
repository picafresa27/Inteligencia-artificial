```python
import numpy as np
#Ejemplo perceptron que evalua si es recomendable comer una comida o si no es recomendable comerla porque es comida chatarra
# Columna 1: ¿Tiene verduras o frutas? (0 = No, 1 = Sí)
# Columna 2: ¿Es comida rapida o con mucha azucar? (0 = No, 1 = Sí)
X = np.array([
    [0, 0],  #Quesadilla -> No tiene verduras ni es comida rapida 
    [1, 0],  #Ensalada -> Tiene verduras, no es comida rapida 
    [0, 1],  #Papas fritas -> No tiene verduras, es comida rapida 
    [1, 1]   #Hamburguesa -> Lleva verduras (lechuga/jitomate), pero es comida rapida
])

#Resultados que obtubimos si es saludable o no
Y = np.array([0, 1, 0, 0])

#Iniciamos los pesos al azar que son como el nivel de importancia o confianza
np.random.seed(42)
pesos = np.random.randn(2)#Peso de 2 para las verduras y la comida rapida o chatarra
sesgo = np.random.randn(1)#Umbral general

#Hiperparámetros
learning_rate = 0.1 #tasa de aprendizaje
epochs = 15 #ciclos o veces que se repetira

print("Datos y pesos inicializados con éxito.")
```

```python
print("Analizando si los platillos son saludables")

#Inicia el entrenamiento
for epoch in range(epochs):#veces que se repetira el ciclo para que aprenda la ia
    errores_cometidos = 0#inicializar los errores en cero
    
    for i in range(len(X)):#recorre la lista de platillos que dimos
        receta = X[i]#saca las caracteristicas de cada platillo
        saludable = Y[i]#analiza la respuesta final, si es saludable o no
        puntuacion = np.dot(receta, pesos) + sesgo#calculo para una puntuacion numerica
        #la ia toma una decision con respecto al calculo anterior
        prediccion = 1 if puntuacion >= 0 else 0
        #calcular que tan equivocado fue el calculo de la ia 
        error = saludable - prediccion
        #ajusta los pesos y sesgo para minimizar el error
        pesos += learning_rate * error * receta
        sesgo += learning_rate * error
        errores_cometidos += abs(error)#cuenta cuantos fallos tubo al finalizar  
    print(f"Ciclo {epoch + 1} - Fallos adivinando si es saludable: {errores_cometidos}")

print("\nFinalizo el entrenamiento de manera correcta")
```

```python
print(f"Pesos finales aprendidos: {pesos}")
print(f"Sesgo final aprendido: {sesgo}")

#Probando el perceptron con resultados de recetas saludables y no saludables
print("\nNuevos platillos")
nuevos_platillos = [
    [0, 1], #Hot dog
    [1, 0], #Caldo de albondigas
    [1, 1]  #Tacos de sada con salsa
]

for receta in nuevos_platillos:
    prediccion = 1 if (np.dot(receta, pesos) + sesgo) >= 0 else 0
    veredicto = "Es saludable :)" if prediccion == 1 else "No es saludable :("
    print(f"platillo [saludable: {receta[0]}, no saludable: {receta[1]}] -> Veredicto de la IA: {veredicto}")
```
