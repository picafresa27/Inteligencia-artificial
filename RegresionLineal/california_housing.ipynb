```python
#librerias
import matplotlib.pyplot as plt #crear graficos y visualizaciones de datos
import numpy as np #calculos y manejo de matrices
import pandas as pd #analisis de datos
from sklearn.datasets import fetch_california_housing # descarga y carga el dataset de viviandas en California
#lo que viene  despues del as es el apodo o nombre corto para hacer mas facil y radipo su llamado

#configura el diseño de las graficas que uasmos con matplotlib
plt.style.use(
    'seaborn-v0_8-whitegrid'
    if 'seaborn-v0_8-whitegrid' in plt.style.available
    else 'default'
)

#cargamos el dataset y con ayuda del as_frame organizamos los datos en una tabla
california = fetch_california_housing(as_frame=True)
df = california.frame #extraemos la tablay la guardamos en la variable df

#con el .value convertimos en un arreglo numerico
X_raw = df.drop(columns=['MedHouseVal']).values #eliminamos de la tabla la columna 'MedHouseVal' que es el precio de las casas y guardamos la matriz en la variable X_raw
Y = df['MedHouseVal'].values #extraemos la columna de 'MedHouseVal' y guardamos en Y

X = (X_raw - np.mean(X_raw, axis=0)) / np.std(X_raw, axis=0) #np.mean para calcular la media o promedio y el np.std calcula la desviacion estandar de la columna
#punto de partida donde la recta cruza el eje, se calcula d emanera automatica
X_b = np.c_[np.ones((X.shape[0], 1)), X] #.shape obtenemos numero de filas, .ones crear columna con numeros 1, .c para pegar la columna al inicio

#saca el numero de casas o registros y cuenta las caracteristicas que tiene cada casa y las imprime
print(f'Registros cargados con exito: {len(X)} con {X.shape[1]} características cada una.')
```

```python
m_features = X_b.shape[1] #cuenta las columnas o caracteristicas que tiene la matriz
theta = np.zeros(m_features) #crea una lista con puros ceros

learning_rate = 0.05 #tasa de aprendizaje
epochs = 1000 #iteraciones que se repetira el ciclo
n = len(Y) #numero de muestras que tendremos

#lista vacia inicialmente, guardaremos el valor del error
historial_costos = []
#bucle que se repetira 1000 veces (valor de epochs)
for epoch in range(epochs):
  Y_pred = X_b.dot(theta)#multiplica las caracteristicas de las casas por los pesos actuales
  cost = (1 / (2 * n)) * np.sum((Y_pred - Y) ** 2)#compara el precio de la ia (Y_pred) contra el precio real de la casa (Y)
  historial_costos.append(cost) #guardamos el error en nuestra lista

  gradients = (1 / n) * X_b.T.dot(Y_pred - Y)#calcula el cambio promedio, hacia donde y con que tanto debe mover el peso para reducir el error
  theta = theta - learning_rate * gradients #nos aseguramos que cambie ligeramente el resultado para la proxima vuelta y no cometer tantos errores

#mensaje de entrenamiento finalizado
print('Entrenamiento finalizado con éxito.')
```

```python
plt.figure(figsize=(9, 4))#creamos ventana donde se va a trazar la grafica
plt.plot(range(epochs), historial_costos, color='darkorange', linewidth=2.5)#agregamos la linea de error
#agregamos titulo a la grafica
plt.title('Evolución del Costo (MSE)', fontsize=11, fontweight='bold', pad=12)
plt.xlabel('Épocas', fontsize=10)#numero de vueltas de entrenamiento
plt.ylabel('Costo', fontsize=10)#indica el tamaño del error
plt.grid(True, linestyle='--', alpha=0.6)#agregamos la cuadricula punteada
#renderizamos la imagen y la mostramos
plt.show()

#calculamos la prediccion final
Y_pred_final = X_b.dot(theta)
plt.figure(figsize=(9, 5))#creamos la vista para la grafica
#dibujamos los puntos deacuerdo a los precios que predijo la ia
plt.scatter(
    Y, Y_pred_final, color='royalblue', alpha=0.4, s=40, label='Predicciones'
)
#dibuja la linea con el valor de la prediccion perfecta
plt.plot(
    [Y.min(), Y.max()],
    [Y.min(), Y.max()],
    color='crimson',
    linestyle='--',
    lw=2.5,
    label='Ideal',
)
#coloca el titulo
plt.title('Valores Reales vs Predichos', fontsize=11, fontweight='bold', pad=12)
#coloca los indicadores de x y y
plt.xlabel('Valor Real', fontsize=10)
plt.ylabel('Valor Predicho', fontsize=10)
plt.legend(fontsize=10)#muestra etiquetas
plt.grid(True, linestyle='--', alpha=0.6)#dibuja lineas punteadas
#renderizamos la imagen y la mostramos
plt.show()
```
